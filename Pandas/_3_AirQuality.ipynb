```python
# Filtering Data
```

```python
data = df[(df["aqi"] > 200) & (df["temperature"] > 16)] ["country"] # (and -> &, not -> ~, or -> |)
data.iloc[0] # This will print as this selects the 0th index of data.
# data.loc[0], this will give an error as this is label based access and the 0th label is not in the data selected.
```

```python
# Queries
```

```python
# Rules
# Expression as a string
# Column based refrence, means we can directly write the names of columns.
# Backticks for col_name with space of special characters
# Operators (&, |, ~, >=, >, <=, <, ==, etc)
# @ to refrence Python variables
# Chained comparisons
```

```python
aqi_val = 200
df.query("aqi > @aqi_val & temperature > 16") ["country"]
```
