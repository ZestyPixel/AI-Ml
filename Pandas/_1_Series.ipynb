```python
# Series
```

```python
s = pd.Series([1, 2, 3, 4, 5])
s
# The numbers on the left are the label, labels are seperate from index. In pandas you can do label or index based axis.
```

```python
print(s[0]) # Index Based
```

```python
s2 = pd.Series([12, 23, 34, 45, 56], index=["Umar", "John", "Rahul", "Bob", "Rob"])
s2
```

```python
print(s2["Umar"]) # Label based
```

```python
print(s2.index)
```

```python
# Series are homogeneous, support vectorised operations, handle missing values with NaN and the size is immutable but the values are mutable
```

```python
s3 = pd.Series([1, 2, 3, 4, 5])
s4 = pd.Series([5, 4, 3, 2, 1])
print(s3 + s4) # Vectorised operation
```

```python
s3[0] = 4 # Changing values sin
s3
```

```python
s5 = s4.drop(0) # To delete an element we have to create a new series since the size of a series is immutable, we cannot drop from the original series
s5
```
