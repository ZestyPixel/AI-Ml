```python
# Cleaning Data
```

```python
import pandas as pd
df = pd.read_csv("raw_data.csv")
df
```

```python
# Cleaning null values
df.isnull() # Can also use isna()
```

```python
df.isnull().sum() # To count no of null values per column
```

```python
df.dropna() # To remove null values containing rows, if you write axis = 1 in the brackets it will drop the columns with null values
```

```python
cleaned_data = df.copy()
cleaned_data["age"] = cleaned_data["age"].fillna(cleaned_data["age"].mean()) #To fill the na values with some particular one.
cleaned_data
```

```python
# ffill() and bfill()
# The basic concept is in ffill() whenever you get a null value, you simply copy the previous value into it(think of it as moving in a top to down direction). 
# The reverse happens in bfill()
```

```python
# Handling duplicate values
df.duplicated() #To check duplicated values
```

```python
df2 = df.copy()
df2.drop_duplicates(inplace = True) #This is to drop duplicate values, inplace makes the actual change in the original data instead of giving a view.
df2
```
