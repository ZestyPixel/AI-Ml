```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
df.sample()
```

```python
# apply()
df2 = df.copy()

df2["tax"] = df2["income"].apply(lambda x: "20%" if x>= 50000 else "10%") # apply() is used to apply a function to values. Here apply is a higher order function meaning
# it takes other functions as arguments.
df2
```

```python
# map()
gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U",}
df2["gender"].map(gender_map)
```

```python
# assign()
df2.assign(new_income = df2["income"] * 1.1)
```

```python
# replace()
df2["country"].replace("USA", "US")
```

```python
# Transforming Data
```

```python
df2
```

```python
df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax"] # Renaming columns, you can change the order to change the order of the columns.
df2.rename(columns={"Income": "Salary"})
df2.rename(index={0: "Zero"})
```

```python
df2["Income"] = df2["Income"].fillna(10000)
df2["Age"] = df2["Age"].fillna(df2["Age"].mean())
df2.sort_values("Income") 
```

```python
df2 = df2.sort_values(["Income", "Age"], ascending = False) # Sorting will happen on basis of income only but if income is ever the same, then age will be the tie breaker
df2
```

```python
# Ranking

df2["Ranking"] = df2["Income"].rank() # Created a ranking based on income. If we pass ascending = False in brackets, it'll make 1 the highest ranking. 
# method = dense for having only whole values. min and max for min or max values in case of tie.
df2
```

```python
# Melt (wide format data to long format) and Pivot(long format to wide format) for Reshaping
# melt(columns we want to keep, ones we want to unpivot/melt into one, new column name, value column name)
df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22],
    "tax": [1, 2, 3, 4]
})

df
```

```python
melted_df = df.melt(["country", "year"], ["sales", "profit", "tax"], "metrics", "value" )
melted_df
```

```python
df2 = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22],
})
melted_df2 = df.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name="metrics",
    value_name="value"
)
melted_df2
```

```python
pivoted_df2 = melted_df2.pivot(index=["country", "year"], 
                 columns="metrics", 
                 values="value")
pivoted_df2
```

```python
# Merging & Joining Data

df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
```

```python
df_customers
```

```python
df_orders
```

```python
# Inner join, only common values in the 'on' column are merged.
pd.merge(df_customers, df_orders, on = "customer_id")
```

```python
# Left join, keep all the values in the left column while put NaN in any missing values
pd.merge(df_customers, df_orders, on = "customer_id", how = "left")
```

```python
# Right join
pd.merge(df_customers, df_orders, on = "customer_id", how = "right")
```

```python
#Keep all, put NaN in any missing values
pd.merge(df_customers, df_orders, on = "customer_id", how = "outer")
```

```python
# Data Concatenation, stacking dataframes on top of each other (rows) or side by side (columns)
```

```python
df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Adam", "Eve", "Bob"]
})

df2 = pd.DataFrame({
    "id": [4, 5, 6],
    "name": ["Charlie", "Fester", "Gilian"]
})
df1
```

```python
df2
```

```python
# row
pd.concat([df1, df2], ignore_index = True) # ignore index to remove the labels which are preserved otherwise and get new labels.
```

```python
pd.concat([df1, df2], ignore_index = True, axis = 1)
```
