```python
# Vectorisation - Normal lists in python are operated upon using loops, but through vectorisation in np arrays, we can apply the
# whole operation on the array in one go
```

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
arr2 = np.array([5, 4, 3, 2, 1])
print(arr**3)
print(arr - 1)
print(arr + arr2)
```

```python
# Braodcasting 
# Instead of copying data to match shapes, NumPy virtually expands the smaller array so element-wise operations can happen efficiently.
# Two arrays are compatible if, starting from the rightmost dimensions:
# Dimensions are equal, OR
# One of them is 1
```

```python
a = np.ones((3, 1))   # (3,1)
b = np.ones((1, 4))   # (1,4)

print(a + b)          # (3,4)
```

```python
# One of the uses of broadcasting is normalisation (fitting values to a certain scale or range)
```

```python
# Normalising using z-score method
arr3 = np.array([[1, 2], [3, 4], [5, 6]])
mean = arr3.mean()
standard_deviation = arr3.std()
normal_array = (arr3-mean)/standard_deviation
print(normal_array)
print(normal_array.mean())
```
