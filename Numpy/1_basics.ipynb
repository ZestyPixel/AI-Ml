# Python Array vs Numpy Array(in terms of speed)

```python
size = 1_000_000 #underscores are just for visual purposes

py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list] 
end = time.time()
print(f"Time taken by list: {end-start}")

np_arr = np.array(py_list)
start_np = time.time()
sq_arr_np = np_arr**2 #vectorisation(applying operation on whole array at once, can be done in np array. In list comprehension like above, it is simply a loop)
end_np = time.time()
print(f"Time taken by numpy array: {end_np-start_np}")
```

```python
print(f"Python list size = {sys.getsizeof(py_list) * len(py_list)} bytes")
print(f"Numpy array size = {np_arr.nbytes} bytes")
```

## Why NumPy Is Faster

- NumPy is implemented in optimized C code
- Stores data in contiguous memory blocks
- Uses less memory (homogeneous data types)
- Supports built-in mathematical functions
- Enables vectorized operations (no Python loops)
- Reduced interpreter and object overhead
