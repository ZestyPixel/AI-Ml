```python
import json
```

```python
#To load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
print(data)
```

```python
print(type(data))
```

### There are many issues in this data. Missing data, inconsistency, duplicates, etc. We have to fix all of this.

```python
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_user = set()

    for user in data:
        # Clean rating
        raw_rating = str(user.get("rating"))

        if raw_rating is not None:
            raw_rating = raw_rating.strip().lower()
            if raw_rating in text_to_num:
                user["rating"] = text_to_num[raw_rating]

        # Handle missing age
        raw_age = user.get("age")
        if raw_age is None:
            user["age"] = None

        # Remove duplicate users
        name = user.get("name", "").strip() # Return empty string if not found

        if name in unique_user:
            continue

        unique_user.add(name)
        cleaned_data.append(user)

    return cleaned_data
```

```python
cleaned_data = clean_data(data)
```

```python
for user in cleaned_data:
    print(user)
```

### Now we have to get meaningful insights from this data like average rating and % of users who gave poor rating(<3)

```python
def insights(cleaned_data):
    avg = 0
    no_of_users = 0
    poor_ratings = 0
    for user in cleaned_data:
        if (float(user["rating"])<3):
            poor_ratings += 1
        avg += float(user["rating"])

    print("Average rating: ", avg/len(cleaned_data))
    print("Percentage of poor ratings: ", (poor_ratings/len(cleaned_data))*100)
```

```python
insights(cleaned_data)
```

### Build a recommendation feature(rule based). If user rating >= 4, recommend same brand products, if not then different brand products

```python
def get_recommendations(data):
    recommendations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if float(user["rating"]) >= 4:
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)

    return recommendations
```

```python
get_recommendations(cleaned_data)
```
