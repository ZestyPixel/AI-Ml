```python
from bs4 import BeautifulSoup
```

```python
with open ("Scraped_Data/Data.html" , "r") as f:
    html_content = f.read()
```

```python
soup = BeautifulSoup(html_content, "lxml")
```

```python
soup.find("h1") # to find the first occurence
```

```python
all_countries = []

all_h3 = soup.find_all("h3")
for h3 in all_h3:
    name = h3.get_text(strip = True) # get_text() to get the content
    population = h3.find_next("div").select("span.country-population")[0].get_text()

    all_countries.append([name, population])
# find_parent(), find_next(), find_previous()
# css selectors: select(), select_one()
# parent, children, descendants, next_sibling(), previous_sibling()
```

```python
import pandas as pd
df = pd.DataFrame(all_countries, columns=["Name", "Population"])
```

```python
df.to_csv("Cleaned_Data/Data.csv", index = False)
```
