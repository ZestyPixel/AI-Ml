```python
import requests
from bs4 import BeautifulSoup
```

```python
URL = "https://quotes.toscrape.com/page/1/"

page_count = 1
while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)
    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("No more valid pages")
        break

    with open(f"Scraped_Data/quotes{page_count}.html", "w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded data from page {page_count}")

    page_count = page_count + 1
```

```python
import os

page_no = 1

while True:
    file_path = f"Scraped_Data/quotes{page_no}.html"

    if not os.path.exists(file_path):
        break

    with open(file_path, "r", encoding="utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")

    all_quotes = soup.select("div.quote")
    life_quotes = []

    for q in all_quotes:
        all_tags = []

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text(strip=True))

        if "life" in all_tags:
            text = q.select_one("span.text").get_text(strip=True)
            author = q.select_one("small.author").get_text(strip=True)

            life_quotes.append([text, author])

    print(life_quotes)

    page_no += 1
```
